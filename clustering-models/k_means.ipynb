# **K Means Clustering**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv(r"C:\Users\LENOVO\Downloads\archive\Mall_Customers.csv",usecols=['Annual Income (k$)','Spending Score (1-100)'])
```

```python
data.sample(5)
```

```python
x = data.to_numpy()
```

```python
plt.figure(figsize=(7,5))
plt.scatter(x[:,0],x[:,1], c='blue', marker='o', edgecolor='black', s=50) 
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.title('Annual Income Vs Spending Score',fontdict=dict(fontsize = 15,fontweight = 'bold'),loc='left')
```

```python
from sklearn.cluster import KMeans
```

```python
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i)
    kmeans.fit_predict(x)
    wcss.append(kmeans.inertia_)
```

```python
wcss
```

## **Elbow Method**

```python
plt.plot(range(1,11),wcss)
```

```python
from sklearn.metrics import silhouette_score
```

## **Silhouette Score**

```python
for k in range(2,11):
    kmeans = KMeans(n_clusters=k,random_state=42)
    labels = kmeans.fit_predict(X=x)
    scrore = silhouette_score(x,labels)
    print(f'K = {k} , silhouette Score = {scrore:.3}')
```

`choosing k = 5(highest value)`

```python
km = KMeans(n_clusters=5)
y_means = km.fit_predict(x)
```

```python
y_means
```

```python
x[y_means==0,1]
```

## **Plotting Clusters**

```python
plt.figure(figsize=(7,5))
plt.scatter(x[y_means==0,0],x[y_means==0,1],alpha=0.6,edgecolors='black')
plt.scatter(x[y_means==1,0],x[y_means==1,1],alpha=0.6,edgecolors='black')
plt.scatter(x[y_means==2,0],x[y_means==2,1],alpha=0.6,edgecolors='black')
plt.scatter(x[y_means==3,0],x[y_means==3,1],alpha=0.6,edgecolors='black')
plt.scatter(x[y_means==4,0],x[y_means==4,1],alpha=0.6,edgecolors='black')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100')
plt.title('K Means Clustering',fontdict=dict(fontsize = 22,fontweight = 'bold'),loc='left')
plt.tight_layout()
```
