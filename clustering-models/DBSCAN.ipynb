## **DBSCEN(Density Based Spatical Clustering of Apllications with Noise)**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data  = pd.read_csv(r"C:\Users\LENOVO\Downloads\Clustering_gmm.csv")
```

```python
data.head()
```

```python
plt.figure(figsize=(20,5))
plt.scatter(data['Weight'],data['Height'],color = 'tomato',marker='s',alpha=0.6)
plt.xlabel('Weight',fontdict= dict(fontsize=11,fontweight='bold'))
plt.ylabel('Height',fontdict= dict(fontsize=11,fontweight='bold'))
plt.title('Height Vs Weight Distribution',fontdict= dict(fontsize=22,fontweight='bold'),pad=15,loc='left')
plt.tight_layout()
plt.show()
```

```python
from sklearn.cluster import DBSCAN
```

```python
db = DBSCAN(eps=1,min_samples=5)
db.fit(data)
```

```python
clusters = db.fit_predict(data)
```

```python
db.labels_
```

```python
np.unique(db.labels_)
```

```python
plt.figure(figsize=(20,5))
plt.scatter(data['Weight'],data['Height'],c=clusters,cmap='viridis', marker = 'o',edgecolors='black')
plt.title("DBSCAN Clustering of Concentric Circles",fontdict= dict(fontsize=22,fontweight='bold'),loc='left')
plt.xlabel('Weight',fontdict= dict(fontsize=11,fontweight='bold'))
plt.ylabel('Height',fontdict= dict(fontsize=11,fontweight='bold'))
plt.show()
```
