## **Series is 1D and DataFrames are 2D objects**

```python
# can we have multiple index 
# Multiindex series (also Known as Hierarchical Indexing)
```

```python
# How to Create multiindex object 
#1) pd.Multiindex.from_tuples()
idx_values = [('cse',2019),('cse',2020),('cse',2021),('cse',2022),('ece',2019),('ece',2020),('ece',2021),('ece',2022)]
multiindex=pd.MultiIndex.from_tuples(idx_values)
print(multiindex.levels[0])
print(multiindex.levels[1])
```

```python
# 2) pd.Multiindex.from_product()
pd.MultiIndex.from_product([['cse','ece'],[2019,2020,2021,2022]])
```

```python
# Creating a series with multiindex object
s = pd.Series([1,2,3,4,5,6,7,8,],index=multiindex)
s
```

```python
# How to Fetch items from such a series
s['cse']
```

```python
# This is multi Index series is 2d because to fetch a particular value we need two values 
```

```python
# unstack : Convert a multi index series into DataFrame
temp = s.unstack()
temp
```

```python
# stack : (unstack opposite)
temp.stack()
```

```python
# Then What is a point of MultiIndex series ?
# ----> Through this same idea we can represent a higher dimension data into lower dimensions (3d -- 2d represent)
```

## **Long And Wide Data**
`Wide`: is where we have single row for every data point with multi columns to hold the values of various attributes .<br>
`Long`: is where for each data point we have as many rows as the number of attributes and each row and columns the value of particular attributes  for a given data point. 


```python
#  melt -- wide to long
#  pivot --- long to wide
```

```python
pd.DataFrame({'cse':[120]}).melt()
```

```python
pd.DataFrame({'cse':[120],'ece':[100],'mech':[50]}).melt(var_name='Brach',value_name='number_students')
```

```python
pd.DataFrame(
    {
        'Branch': ['cse','ece','mech'],
        '2020':[100,150,60],
        '2021':[120,130,80],
        '2022':[130,140,70]
    }
).melt(id_vars=['Branch'],var_name='year',value_name='students')
```
