```python
# drop_duplicates : drop duplicates values
temp = [1,2,1,2,2,3,5,4,9]
pd.Series(temp).drop_duplicates(keep='last') # drop duplicates and keep last
```

```python
pd.Series(temp).duplicated().sum()
```

```python
temp = pd.Series([1,2,3,np.nan,5,6,np.nan,8,np.nan])
print(temp.isnull().sum())
```

```python
# dropna : drop nan values
temp.dropna()
```

```python
# fillna : to fill missing values 
temp.fillna(temp.mean())
```
