## **Pandas**

## **Pandas Series**
- like a column in a table.
- It is 1-D array holding data of any type

```python
import numpy as np
import pandas as pd
```

### **Series from list**

```python
# string
country = ['India','Pakistan','USA',"Nepal",'SriLanka']
pd.Series(country)
```

```python
# Integers
runs = [45,67,37,86,98,54]
pd.Series(runs)
```

```python
# custom index
marks = [73,89,95,98,92,88]
subjects = ['Maths','Physics','Chemistry','Hindi','English','Physical Education']
x=pd.Series(marks,index=subjects)
x
```

```python
# setting a name
marks= pd.Series(marks,index=subjects,name='Dheeraj Marks')
marks
```

## **Series from dict**

```python
marks = {
    'maths':73,
    'chemistry':95,
    'Physics':89 
}
marks_series = pd.Series(marks,name='Dheeraj_Marks')
marks_series
```

## **Series Attributes**

```python
# Size : Total Values
marks_series.size
```

```python
# dtype :  items data type
marks_series.dtype
```

```python
# name : name of series
marks_series.name
```

```python
# is_unique : return True if values are non identical esle False
marks_series.is_unique
```

```python
# index : return index object(where index values are stored)
marks_series.index
```

```python
# values : return values 
marks_series.values
```
