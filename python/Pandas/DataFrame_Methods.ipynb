```python
point_df = pd.DataFrame({
    'Point1' : [(3,4),(4,5),(0,0),(3,1),(6,8)],
    'Point2': [(-3,4),(6,3),(4,2),(2,5),(5,6)]
})
point_df
```

```python
point_df['Point1'][0]
```

```python
def euclidien_distance(row):
    pt_1 = row['Point1']
    pt_2 = row['Point2']
    return ((pt_1[0]-pt_2[0])**2 + (pt_1[1]-pt_2[1])**2)**0.5
```

```python
point_df['distance'] = point_df.apply(euclidien_distance,axis=1)
point_df
```
