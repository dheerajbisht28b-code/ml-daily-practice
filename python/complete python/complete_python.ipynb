```python
def is_even(x):
    """
    this function is used to determine number is even or not
      """
    if type(x) == int:
         if(x%2==0):
            return f"{x} is even."
         else:
            return f"{x} is not even."
    else:
        return f"{x} is {type(x)},Enter valid data type."
   
is_even(9998)## calling a function.
for i in range(1,11):
    x = is_even(i)
    print(x)
# when we create a function we pass value called paremeter.but when we call function by given value that called argument.
```

```python
# args and kwargs : special python keywords that are used to pass the variable length of arguments to a function.
# args = allows us to pass a variable number of non-keyword arguments to a function.
# kitne bhi inputs de sakte hai.
def multiply(*args): # we can use any word in place of args like *mul, * m etc.
    product = 1
    for i in args:
        product = product*i
    return product
```

```python
multiply(2,3,4,5)
```

```python
# **kwargs: allows us to pass any number of key number of keyword arguments.
def display(**kwargs):
    for (key,value) in kwargs.items():
        print(key,'->',value)
```

```python
def find_max(*args):

    for i in args:
        if not isinstance(i,(int,float)):
            return "only numbers are allowed"
    max_num = args[0]
    for num in args:
        if num > max_num:
            max_num = num
    return max_num
```

```python
find_max(1,2,33,12,23,57)
```

## **Lambda Function**
- A lambda function is a small anonymous(no name) functions.

```python
# SyntaxX: lambda a,b : a+b
```

```python
a = lambda x : x**2
print(a(2))
print(a(4))
```

```python
a = lambda x,y: x+y
print(a(3,4))
print(a(99,56))
```

```python
""" diff between lambda vs normal function
- no name
- lambda has no return value(infact returns a function)
- lambda is written in 1 line
- not reusable
why we use lambbda function
They are used with HOF(higher order functions)
"""
```

```python
# check if string has 'a'
a = lambda s : 'a'in s
s = "hello"
print(a(s))
```

```python
#check odd or even 
a = lambda x : "even" if(x%2 == 0) else "odd"
print(a(3))
print(a(4))
```

###  **Higher order Functions**
- A function that:
1) Takes another function as argument, or
2) Return a function as result,or
3) Both

```python
# example
def square(x):
    return x**2
def transform(f,l): # transform is higher order function
    output = []
    for i in l:
        output.append(f(i))
    print(output)
l = [1,2,3,4,5]
transform(square,l)
# rather than creating different - difffrent functions we can use lambda function.
transform(lambda x: x**3,l)
```

### **Map**
syntaxx : map(function,iterable)<br>
Built in function used to apply a function to a every item of a iterable(like, list,tuples)

```python
# square the items of a list
list(map(lambda x: x**2,[2,3,4]))
```

## **filter()**
syntax : filter(function,iterable)<br>
use : it returns  an iterater containing only the elements that satisfy a condition.

```python
#numbers greater than 5 
L = [3,4,5,6,7]
list(filter(lambda x: x>5,L))
```

```python
# fetch fruits starting from 'a'
L= ["apple",'orange','mango']
list(filter(lambda x: x.startswith('a'),L))
```

## **reduce()**<br>
use; repaetedly appliies a function to the elements of an iterable and reduce them in single value.<br>
avilable in functools module

```python
# sum of all 

''' 
x = 1,y=2 , x+y = 3
L = [3,3,4]
x = 3 , y =3 x+y =6
L = [6,4]
x = 6, y= 4 x+y = 10(result)
'''
import functools
functools.reduce(lambda x,y: x+y,[1,2,3,4])
```

```python
# find min
functools.reduce(lambda x,y: x if(x<y) else y,[199,2,34,300,1])
```

map() = transform every element<br>
filter() = keeps elements matching condition<br>
reduce() = combines all elements into single value
