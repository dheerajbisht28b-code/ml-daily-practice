##  **Using Context Manager(with)**
- it's a good idea to close a file after usage as it will free up the resources
- if we dont close it , garbage collecter would  close it
- with keyword closes the file as soon as usage is over

```python
# with
with open('sample.txt', 'w') as f:
    f.write('selmon bhai')
```

```python
# try f.read() now
with open('sample.txt','r') as f:
    print(f.read())
```

```python
with open('sample.txt','r') as f:
    print(f.read(10))
    print(f.read(10))
```

```python
# Benifit ? --> to load a big file in memory
big_l = ['hello world\n' for i in range(1000)]

with open('big.txt','w') as f:
    f.writelines(big_l)
```

```python
with open('big.txt','r') as f:
    chunk_size = 10

    while( len(f.read(chunk_size)) >0):
        print(f.read(chunk_size),end='*')
        f.read(chunk_size)
```

```python
# seek and tell function
with open('big.txt','r') as f:
    print(f.read(10))
    print(f.tell()) # how many chars we have processed
```

```python
with open('sample.txt','r') as f:
    print(f.read(10))
    print(f.seek(0)) # we return to o position print from start, we can change acc to context
    print(f.read(10))
```

```python
# seek during write
with open('sample.txt','w') as f:
    f.write('Hello')
    f.seek(0)
    f.write("Xa")
```

# **Serialization and Deserialization**
- **Serialization**: process of converting python data type to JSON format
- **Deserialization** Process to converting JSON to Python data types
- Json (Java script on notation): universal text format understood by all the programming languages.

```python
import json
L = [1,2,3,4]

with open('demo.txt',"w") as f:
    json.dump(L,f)
```

```python
# dict
d= {
    'name':'Dheeraj',
    'age':18,
    'gender': 'male'
}
with open('sample.txt','w') as f:
    json.dump(d,f,indent=4)
```

```python
# deserialization
import json
with open('sample.txt','r') as f:
    d = json.load(f)
    print(d)
    print(type(d))
```

```python
# serialize and deserialize tuple
import json 
t = (1,2,3,4,9)
with open('demo.json','w') as f:
    json.dump(t,f)
```

```python
# serialize and deserialize a nested dict
d = {
    'name': 'Dheeraj',
    'marks': [35,56,78,42,56]
}
with open('demo.json','w') as f:
    json.dump(d,f)
```

# **Serializing and Deserializing custom objects**

```python
class Person:

    def __init__(self,fname,lname,age,gender):
        self.fname = fname
        self.lname = lname
        self.age = age
        self.gender = gender
# format to printed in 
# Dheeraj Singh age --> 33, gender --> male
```

```python
person = Person('Dheeraj','Singh',33,'Male')
```

```python
import json

def show_object(person):
    if (isinstance(person,Person)):
        return {'Name': person.fname + " " + person.lname ,'age': person.age,'gender': person.gender}
with open('demo.json','w') as f:
    json.dump(person,f,default=show_object,indent=4)
```

```python
# deserialize 
with open('demo.json','r') as f:
    d = json.load(f)
    print(d)
```

## **Pickiling**
`pickling`: is the process whereby a python object hierarchy is converted into a byte stream, and<br>
`unpickling`: is the inverse operation , where by a byte stream (from a binary file or bytes -like objects)<br> is converted back into object hierarchy  

```python
class Person:

    def __init__(self,name,age):
        self.name = name
        self.age = age

    def display_info(self):
        print(f'Hi my name is {self.name} and I am {self.age} years old')
```

```python
p = Person('Dheeraj',18)
```

```python
# Pickle dump
import pickle
with open('person.pkl','wb') as f:
    pickle.dump(p,f)
```

```python
import pickle
with open('person.pkl','rb') as f:
    d=pickle.load(f)
d.display_info()
```

Pickle vs Json:
- Pickle lets the user to store data in binary format . Json let the user store data in a human readable text foramt
