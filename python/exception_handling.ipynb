```python
# Catching specific exception
try:
    m=3
    f = open('sample.txt','r')
    print(f.read())
    print(m)
    print(5/2)
    l=[1,2,3,4]
    l[100] 
except FileNotFoundError:
    print('file not found')
except NameError:
    print('Variable not defined')
except ZeroDivisionError:
    print("can't divide by zero!!")
except Exception as e:
    print(e)
```

```python
# else
try:
    f = open('sample.txt','r') 
except FileNotFoundError:
    print('File not found')
except Exception:
    print('something wronge!!!')
else:
    print(f.read())
```

```python
class Bank:

    def __init__(self,balance):
        self.balance = balance

    def withdraw(self,amount):
        if amount < 0:
            raise Exception('amount cannot be negative!!!')
        if amount>self.balance:
            raise Exception('Not Enough Balance!!!')
        self.balance -= amount

obj = Bank(10000)
try:
    obj.withdraw(300000)
except Exception as e:
    print(e)
else:
    print('Withdraw Sucessfully!!!')
    print('current balance ',obj.balance)
```

```python
# Costum Exception
class MyException(Exception):
    def __init__(self, massage):
        print(massage)
class Bank:

    def __init__(self,balance):
        self.balance = balance

    def withdraw(self,amount):
        if amount < 0:
            raise MyException('amount cannot be negative!!!')
        if amount>self.balance:
            raise MyException('Not Enough Balance!!!')
        self.balance -= amount

obj = Bank(10000)
try:
    obj.withdraw(-300000)
except MyException as e:
    pass
else:
    print('Withdraw Sucessfully!!!')
    print('current balance ',obj.balance)
# custom classes are helpfull for full control and application based error......
        
```
