## **Super Keyword**
- **Super() is an inbuilt function which is used to invoke the parent class methods and constructor**
- Super cannot access attributes 
- super cannot used outside the class
- super is used inside  the child class

```python
# Method Overriding
class Phone:
    def __init__(self,price,brand,camara):
        print('Inside phone constructor')
        self.price = price
        self.brand = brand
        self.camara = camara

    def buy(self):
        print('buying a phone')

class SmartPhone(Phone):
    
    def buy(self):
        print('buying a smartphone')
        # syntax to call parent's buy method 
        super().buy()
    
s = SmartPhone(2000,'Apple',50)
s.buy()
```

```python
# super keyword call inside the class , we cannot call it  from outside of class
# s.super.buy() 
```

```python
# super --> constructor
class Phone:
    def __init__(self,price,brand,camara):
        print('Inside phone constructor')
        self.price = price
        self.brand = brand
        self.camara = camara

    def buy(self):
        print('buying a phone')

class SmartPhone(Phone):
    def __init__(self, price, brand, camara,os,ram):
        print('Inside Smartphone constructor')
    
        super().__init__(price, brand, camara)
        self.os = os
        self.ram = ram
        print('Inside Smartphone constructor')
    
s = SmartPhone(20000,'Samsung',50,'Android',4)
print(s.os)
print(s.brand)
# Through super method we can access the attributes and methods of parent class
        
```

### **Types of inheritance**
- Single
- Multilevel
- Hierarchical
- Multiple(Diamond Problem)
- Hybrid 

```python
# Single
class Phone:
    def __init__(self,price,brand,camara):
        print('Inside phone constructor')
        self.price = price
        self.brand = brand
        self.camara = camara

    def buy(self):
        print('buying a phone')

class SmartPhone(Phone):
    def buy(self):
        print('buying a smartphone')
```

```python
#  Multilevel
class Product:
    def review(self):
        print('Product customer review')
class Phone(Product):
    def __init__(self,price,brand,camara):
        print('Inside Phone Constructor')
        self.price = price
        self.brand = brand
        self.camara = camara

    def buy(self):
        print('Buying a phone')

class SmartPhone(Phone):
    pass

s = SmartPhone(2000,'Apple',12)
s.buy()
s.review()
```

```python
#Hierarchical
class Phone:
    def __init__(self,price,brand,camara):
        print('Inside Phone Constructor')
        self.price = price
        self.brand = brand
        self.camara = camara

    def buy(self):
        print('Buying a phone')

class SmartPhone(Phone):
    pass
class FeaturePhone(Phone):
    pass
SmartPhone(1000,'Apple',"13px").buy()
FeaturePhone(10,'Lava','1px').buy()
```

```python
# Multiple
class Phone:
    def __init__(self,price,brand,camara):
        print('Inside Phone Constructor')
        self.price = price
        self.brand = brand
        self.camara = camara

    def buy(self):
        print('Buying a phone')
        
class Product:
    def review(self):
        print('Product customer review')

    def buy(self):
        print('buy a product')
# Method resolution order(MRO)
class SmartPhone(Phone,Product): #  Phone class inherit first than Product class 
    pass

s = SmartPhone(2000,'Apple',12)
s.buy() # Phone class buy is execute (because we inherit Phone calss first)
```

## **Abstraction**

```python
from abc import ABC,abstractmethod
class BankApp(ABC):

    def database(self):
        print('Connected to database')

    @abstractmethod
    def security(self):
        pass
    @abstractmethod
    def display(self):
        pass
```

```python
class MobileApp(BankApp):

    def mobile_login(self):
        print('login into mobile')

    def security(self):
        print('mobile security')

    def display(self):
        print('None')
```

```python
mob = MobileApp()
mob.security()
```
