## **Scope and LEGB Rule**
- A scope is a textual region of a python Program where a namespace is dfirectly accessible.<br>
- The interpreter seraches for name from the inside out, looking in the local, enclosing , global and finally the built in scope .<br> if the interpreter doesn't find the name in any of these locations, then Python Raises a NameError Exception.

```python
# local and global

a = 2 # Global Variable
def temp():
    b=3 # Local Variable
    print(b)
temp()
print(a) 
```

```python
# local and global --> same name

a = 2 
def temp():
    a=3 
    print(a)
temp()
print(a) 
```

```python
# local and global --> local does not have but Global have (LEGB rule)
a = 2 
def temp():
    print(a)
temp()
print(a) 
```

```python
# local and global --> global created inside local
def temp():
    global a
    a=1
    print(a)  
temp()
print(a) 
```

```python
# local and global --> function paremeter is local

a = 2 
def temp(b):
    b=3 
    print(b)
temp(a)
print(a) 
```

```python
# Enclosing scope (nested)
a=6
def  outer(): #enclosing/non local
    a = 3
    def inner(): #local
        a=4
        print(a)
    inner()
    print(a)
outer()
print(a) # global
```

## **Decorators**
- A decorator in a python is a function that recieves another function as input and adds some functionality(decoration) to it and returns it.
- This can happen only because python functions are 1st class citizens.<br>
- There are 2 types of decorators available in python
1) Built-in decorators like `@staticmethod`,`@classmethod`,`@abstractmethod` and `@property` etc
2) User defined decorators that we programmers can create according to our needs

```python
# Python are 1st class function
def modify(func,num):
    return func(num)
def square(num):
    return num**2
modify(square,2)
```

```python
# simple example
def my_decorator(fun):
    def wrapper():
        print('*******************')
        fun()
        print('*******************')
    return wrapper
def hello():
    print('Hello')
a= my_decorator(hello)
a()
def display():
    print('Hello SyrnaxX')
b= my_decorator(display)
b()
```

```python
def outer():
    a = 3
    def inner():
        print(a)
    return inner
b= outer()
b()
```

```python
# conclusion: After destroying the outer function inner function can access the outer function
```

```python
def my_decorator(fun):
    def wrapper():
        print('*******************')
        fun()
        print('*******************')
    return wrapper
@my_decorator
def hello():
    print('Hello')
hello()
```

```python
# meaningfull example:
import time
def timer(fun):
    def wrapper(*args):
        start = time.time()
        fun(*args)
        print(f'time taken by {fun.__name__}:{time.time()-start} sec ')
    return wrapper
@timer
def hello():
    print('Hello World')
    time.sleep(2)
hello()
@timer
def display():
    print('display something........')
    time.sleep(4)
display()
# But this will work only for 1 type function (that takes no input)
```

```python
@timer
def yes(num,num1):
    time.sleep(1)
    print(num*num1*2)
yes(3,33)
 # this handleby *args
```

```python
def sanity_check(data_type):
    def outer_wrapper(func):
        def inner_wrapper(*args):
            if (type(*args) == data_type):
                func(*args)
            else:
                raise TypeError('This is not valid data type....')
        return inner_wrapper
    return outer_wrapper
```

```python
# A big Problem
@sanity_check(int)
def square(num):
    print(num**2)
square(9)
```

```python
@sanity_check(str)
def greet(name):
    print("Wellcome",name)
greet('SyrnaxX')
```
