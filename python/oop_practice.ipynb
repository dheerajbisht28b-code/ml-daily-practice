# **Function Class**

```python
class Fraction:
    # parameterized constructor
    def __init__(self,x,y):
        self.numerator = x
        self.denominator = y
    
    def __str__(self): # second method function (How show this object look when printed)
        return '{}/{}'.format(self.numerator,self.denominator)
        #f'{self.numerator}/{self.denominator}' #or
    
    def __add__(self, other):
        new_num = self.numerator *other.denominator + other.numerator * self.denominator
        new_deno = self.denominator*other.denominator
        return f'{new_num}/{new_deno}'
    
    def __sub__(self, other):
        new_num = self.numerator *other.denominator - other.numerator * self.denominator
        new_deno = self.denominator*other.denominator
        return f'{new_num}/{new_deno}'
    
    def __mul__(self, other):
        new_num = self.numerator *other.numerator
        new_deno = self.denominator*other.denominator
        return f'{new_num}/{new_deno}'
    
    def __truediv__(self, other):
        new_num = self.numerator *other.denominator
        new_deno = self.denominator*other.numerator
        return f'{new_num}/{new_deno}'
        

```

```python
a = Fraction(2,3)
b =Fraction(4,5)
print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

# **Co-ordinate Class**

```python
class Point():

    def __init__(self,x,y):
        self.x_cod = x
        self.y_cod = y

    def __str__(self):
        return f'({self.x_cod},{self.y_cod})'
    
    def euclidien_distance(self,other):
        X = (other.x_cod-self.x_cod)**2
        Y = (other.y_cod-self.y_cod)**2
        return f"{(X+Y)**0.5:.2f} units"
    
    def distance_from_origin(self):
        return self.euclidien_distance(Point(0,0))
        
```
