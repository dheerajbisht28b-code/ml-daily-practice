```python
# 2 WAP a program to calculate factorial of a number
a  = int(input("enter the number: "))
i  =1
fact = 1
while(i<=a):
    fact *=i
    i+=1
print(f"{a}! = {fact}")
```

```python
# 4 write a programme to reverse a given string
text = input("Enter a sentance....")
s = text.lower()
result = []
for i in s.split():
    result.append(i)
    
print(str(result[::-1]))
```

```python
# 5 check no. is palindrome
num = int(input("enter a number"))
real= num
result = 0
while(num>0):
    rem = num%10
    result = result*10 + rem
    num//=10
if(result == real):
    print(f'{real} is palindrome')
else:
    print(f"{real} is not palindrome")
```

```python
# 7 Create a number guessing number
import random
jackpot = random.randint(1,100)
guess = int(input("Guess the jackpot number: "))
count=1
while(jackpot!=guess):
    if(guess>jackpot):
        print('guess small...')
    else:
        print('guess large....')
    count+=1
    guess = int(input("Guess the jackpot number: "))
print(f"Correct guess!!!!!\nJackpot number is {jackpot}")
print(f"You guess correct number in {count} attempts")
```

```python
#8 WAP to print prime number in a given range 
lower = int(input("Enter the lower range: "))
upper = int(input("Enter the upper range: "))
for i in range(lower,upper+1):
    for j in range(2,i):
        if(i%j==0):
            break
    else:
        print(i,end= " ")
```
